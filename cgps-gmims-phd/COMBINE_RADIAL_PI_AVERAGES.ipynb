```python
fieldnames = ['ob02','ob03','ob04','ob05',
              'ob08','ob09','ob10','ob11',
              'ob12','ob13','ob14','ob15',
              'ob16','ob17','sn01','sn02',
              'sn03','rr22']
numname = np.size(fieldnames)

fieldandband = []

for i in range(0,numname):
    fieldandband.append(fieldnames[i]+'_aq')
    fieldandband.append(fieldnames[i]+'_au')
    fieldandband.append(fieldnames[i]+'_bq')
    fieldandband.append(fieldnames[i]+'_bu')
    fieldandband.append(fieldnames[i]+'_cq')
    fieldandband.append(fieldnames[i]+'_cu')
    fieldandband.append(fieldnames[i]+'_dq')
    fieldandband.append(fieldnames[i]+'_du')

print(fieldandband)

numfields = np.size(fieldandband)
print(numfields)
```

```python
C_PI_files = []
G_PI_files = []

for i in range(0,numfields):
    
    PI_files = ['../../GMIMS_CGPS_OB_RW/THESIS_pics_imageplane/C_initial_image_'+fieldandband[i]+'.fits',
                '../../GMIMS_CGPS_OB_RW/THESIS_pics_imageplane/G_initial_image_'+fieldandband[i]+'.fits']
    
    field = fits.open(PI_files[0])
    hdr = field[0].header
    data = field[0].data[0,0,:,:]
    C_PI_files.append(data)
    
    field = fits.open(PI_files[1])
    hdr = field[0].header
    data = field[0].data[0,0,:,:]
    G_PI_files.append(data)
```

```python
for i in range(0,numfields):
    PI_avrg_C = np.nanmean(abs(C_PI_files[i]))
    print(PI_avrg_C)
    
print('==================================')
for i in range(0,numfields):
    PI_avrg_G = np.nanmean(abs(G_PI_files[i]))
    print(PI_avrg_G)
```
