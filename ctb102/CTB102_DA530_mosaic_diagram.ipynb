```python
import numpy as np
import matplotlib.pyplot as plt
import math
from matplotlib.pyplot import cm
from cycler import cycler
plt.rcParams['axes.prop_cycle'] = cycler(color='bgrcmyk')
```

```python
drao_fields = open('drao_field_names.ann','r')
lines=drao_fields.readlines()

lon_drao  = []
lat_drao  = []
name_drao = []
for x in lines:   
    lon_drao.append(x.split()[2])
    lat_drao.append(x.split()[3])
    name_drao.append(x.split()[4])
drao_fields.close()

for i in range(0,len(lines)):
    print(name_drao[i],float(lon_drao[i]),float(lat_drao[i]),1)
```

```python
mos_names = ['MCH1','MCH2','MCH3','MCH4',
             'MH2', 'MIJ2','MK2', 'ML2']
mos_lon = [100.75, 96.75, 92.75, 88.75,
           100.75, 96.75, 92.75, 88.75]
mos_lat = [ 7., 7., 7., 7.,
            3., 3., 3., 3.]
r = 75./60.
d = 0.001
width = 5.12
# Fraction of the width corresponding to the maximum distance
# from the mosaic centre for which to include a field label.
# a = 0.5 means only fields that are at least halfway inside
# the mosaic are included, bigger numbers mean fields that are
# less than halfway in are also included:
a = 0.65
#a = 0.7
#=================
fs=16
#================


centres = open('/home/ordoga/Python/cgps-gmims/ch_fields.dat','r')
lines=centres.readlines()
centres_col_num = 1

c_lon = []
c_lat = []
name = []
color_list =[]

for x in lines:
    name.append(x.split()[0])    
    c_lon.append(x.split()[1])
    c_lat.append(x.split()[2])
    color_list.append(x.split()[3])
centres.close()

num = np.size(name)
```

```python
c_lon = np.asfarray(c_lon)
c_lat = np.asfarray(c_lat)
print(c_lon)
```

```python
fig = plt.figure(figsize=(20,10))
plt.subplots_adjust(top = 0.95, bottom = 0.05, right = 0.95, left = 0.05, hspace=0.2, wspace=0.2)

actual_colors = ['C1','C0','C2']

for k in range(241,249):
    
    j = k-241    
    cl = mos_lon[j]
    cb = mos_lat[j]
    mos_name = mos_names[j]
    
    ax = fig.add_subplot(k,aspect='equal')
    for i in range(0,num):
        xcirc = np.arange(-r,r+d,d)
        ycircp = np.sqrt(r**2-xcirc**2)
        ycircn = -np.sqrt(r**2-xcirc**2)

        ax.plot(xcirc+float(c_lon[i]),ycircp+float(c_lat[i]),color=actual_colors[int(color_list[i])],
                linewidth=2,zorder=0)
        ax.plot(xcirc+float(c_lon[i]),ycircn+float(c_lat[i]),color=actual_colors[int(color_list[i])],
                linewidth=2,zorder=0)
        
        if ( (abs(float(c_lon[i])-cl) < a*width) & (abs(float(c_lat[i])-cb) < a*width) ):
            ll_loc = float(c_lon[i])+0.2
            bl_loc = float(c_lat[i])-0.1
            if float(c_lon[i]) > cl+0.47*width:
                ll_loc = cl+0.47*width
            if float(c_lon[i]) < cl-0.37*width:
                ll_loc = cl-0.37*width
            if float(c_lat[i]) > cb+0.46*width:
                bl_loc = cb+0.45*width
            if float(c_lat[i]) < cb-0.48*width:
                bl_loc = cb-0.48*width
            plt.text(ll_loc,bl_loc,str(name[i]),fontsize=fs,fontweight=550)
    ax.tick_params(axis="x", labelsize=fs)
    ax.tick_params(axis="y", labelsize=fs)
    if (j==0 or j==4):
        ax.set_ylabel('Galactic Latitude (degrees)',fontsize=fs)
    if (j>3):
        ax.set_xlabel('Galactic Longitude (degrees)',fontsize=fs)
    ax.set_xlim(cl+width/2.,cl-width/2.)
    ax.set_ylim(cb-width/2.,cb+width/2.)
    ax.set_title(mos_name,fontsize=fs+4)
    ax.grid() 
    
plt.savefig('CTB102_DA530_mosaics.pdf')
```
